## 문제 1. 올림

상품 `items`개를 한 상자에 `per_box`개씩 담습니다. 필요한 상자 수를 `boxes`에 넣습니다. `math`를 import하고, `items / per_box`를 `math.ceil`로 올림합니다.

````python
items = 47
per_box = 10
import math
boxes = math.ceil(items / per_box)
````

````python
# 확인 칸
assert "boxes" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert boxes != 4, "items // per_box는 4지만 7개가 남습니다. 올림(math.ceil)을 씁니다"
assert boxes == 5, "필요한 상자는 5개입니다"
````

## 문제 2. 로또

`random.sample`로 1부터 45까지(45 포함) 번호 6개를 **겹치지 않게** 뽑고, `sorted`로 작은 순서로 정렬한 리스트를 `lotto`에 넣습니다.

````python
import random
lotto = sorted(random.sample(range(1, 46), 6))
````

````python
# 확인 칸
assert "lotto" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert lotto is not None, "sort()는 None을 돌려줍니다. 정렬한 새 리스트는 sorted()로 만듭니다"
assert type(lotto) == list and len(lotto) == 6, "번호 6개를 담은 리스트여야 합니다"
assert len(set(lotto)) == 6, "번호가 겹쳤습니다. random.sample을 씁니다"
assert all(1 <= n <= 45 for n in lotto), "번호는 1부터 45까지입니다. range(1, 46)"
assert lotto == sorted(lotto), "작은 순서로 정렬합니다"
````

## 문제 3. 며칠 남았나

`datetime`의 `date`로 `today`를 2026년 10월 27일로, `deadline`을 2026년 11월 2일로 만들고, `deadline`에서 `today`를 뺀 날 수를 정수로 `days_left`에 넣습니다.

````python
from datetime import date
today = date(2026, 10, 27)
deadline = date(2026, 11, 2)
days_left = (deadline - today).days
````

````python
# 확인 칸
assert "days_left" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(days_left) == int, "날짜끼리 뺀 결과에 .days를 붙이면 정수가 됩니다"
assert days_left == 6, "10월 27일부터 11월 2일까지 6일입니다"
````

## 문제 4. 날짜 모양 바꾸기

`date(2026, 10, 27)`을 `strftime`으로 `"2026/10/27"` 모양의 문자열로 바꿔 `text`에 넣습니다.

````python
from datetime import date
text = date(2026, 10, 27).strftime("%Y/%m/%d")
````

````python
# 확인 칸
assert "text" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(text) == str, "strftime은 문자열을 돌려줍니다"
assert text == "2026/10/27", "%Y/%m/%d 모양입니다"
````
