## 문제 1. 짝수 내주기

0부터 `n`까지(포함) 짝수를 하나씩 `yield`하는 제너레이터 함수를 완성합니다.

```python
# 풀어 쓴 모양: 0부터 n까지 하나씩 보고 짝수일 때만 yield
def even_numbers(n: int):
    for i in range(n + 1):      # n까지 포함하려고 n + 1
        if i % 2 == 0:          # 2로 나눈 나머지가 0이면 짝수
            yield i             # 하나 내주고 여기서 멈췄다가, 다음 값을 달라고 하면 이어서


# range 간격으로 줄인 모양: 0부터 2씩 건너뛰면 짝수만 나온다
def even_numbers(n: int):
    for i in range(0, n + 1, 2):
        yield i
```

```python
# 써 보기
print(list(even_numbers(6)))  # => [0, 2, 4, 6]
```

```python
# 확인 칸
assert "even_numbers" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
gen = even_numbers(6)
assert gen is not None and hasattr(gen, "__next__"), "return 대신 yield를 써서 제너레이터를 만듭니다"
assert list(gen) == [0, 2, 4, 6], "0부터 6까지(포함) 짝수는 0, 2, 4, 6입니다"
assert list(even_numbers(1)) == [0], "n이 1이면 0 하나입니다"
```

## 문제 2. 숫자만 골라 내주기

문자열 리스트에서 정수로 바꿀 수 있는 값만 `int`로 바꿔 하나씩 `yield`하는 제너레이터 함수를 완성합니다.

```python
def only_numbers(texts: list[str]):
    for t in texts:
        try:
            yield int(t)        # 바꿀 수 있으면 int로 바꿔 하나 내준다
        except ValueError:
            pass                # "둘" · ""처럼 바꿀 수 없으면 내주지 않고 다음 값으로
```

```python
# 써 보기
print(list(only_numbers(["1", "둘", "3", ""])))  # => [1, 3]
```

```python
# 확인 칸
assert "only_numbers" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
gen = only_numbers(["1", "둘", "3", ""])
assert gen is not None and hasattr(gen, "__next__"), "yield를 써서 제너레이터를 만듭니다"
try:
    result = list(gen)
except ValueError:
    assert False, "\"둘\" · \"\"은 int로 바꿀 수 없어 ValueError가 납니다. int(…)를 try로 감싸고 except ValueError에서 넘어갑니다"
assert result == [1, 3], "바꿀 수 있는 \"1\"과 \"3\"만 int로 내줍니다. try · except ValueError"
```

## 문제 3. 제곱의 합

1부터 `n`까지 수의 제곱의 합을 돌려주는 함수를 **제너레이터 표현식** 한 줄로 완성합니다.

```python
# 풀어 쓴 모양: 합을 담을 변수에 제곱을 하나씩 더한다
def square_sum(n: int) -> int:
    total = 0
    for i in range(1, n + 1):   # 1부터 n까지(포함)
        total += i * i
    return total


# 리스트 컴프리헨션 모양: 제곱 리스트를 다 만든 뒤 sum
def square_sum(n: int) -> int:
    return sum([i * i for i in range(1, n + 1)])


# 제너레이터 표현식 모양(문제가 바라는 답): 대괄호를 빼면 리스트를 만들지 않고 하나씩 sum에 건넨다
def square_sum(n: int) -> int:
    return sum(i * i for i in range(1, n + 1))
```

```python
# 써 보기
print(square_sum(3), square_sum(10))  # => 14 385
```

```python
# 확인 칸
assert "square_sum" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert square_sum(3) is not None, "print가 아니라 return으로 돌려줍니다"
assert square_sum(3) == 14, "1 + 4 + 9 = 14"
assert square_sum(10) == 385, "1부터 10까지 제곱의 합은 385입니다"
```

## 문제 4. 대문자 데코레이터

감싼 함수가 돌려주는 문자열을 대문자로 바꿔 돌려주는 데코레이터 `upper`를 완성합니다.

```python
# 풀어 쓴 모양: 원래 함수를 부르고, 받은 결과를 바꾼 뒤 돌려준다
def upper(func):
    def wrapper(*args, **kwargs):           # 감쌀 함수가 무엇을 받든 그대로 받는다
        result = func(*args, **kwargs)      # 원래 함수를 부른다 → "hello, sesac"
        return result.upper()               # 대문자로 바꿔 돌려준다 → "HELLO, SESAC"
    return wrapper                          # wrapper를 부르지 않고 함수 자체를 돌려준다


# 줄인 모양: 부르고 바로 .upper()
def upper(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return wrapper
```

```python
# 써 보기
@upper
def hello(name):
    return f"hello, {name}"


print(hello("sesac"))  # => HELLO, SESAC
```

```python
# 확인 칸
assert "upper" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"


@upper
def hello(name):
    return f"hello, {name}"


assert hello is not None, "데코레이터는 wrapper 함수를 return합니다"
assert hello("sesac") is not None, "wrapper는 대문자로 바꾼 문자열을 return합니다"
assert hello("sesac") == "HELLO, SESAC", "hello(\"sesac\")는 \"HELLO, SESAC\"입니다"
```
