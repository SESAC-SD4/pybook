# 07 연습

1. 빈 칸의 함수 모양을 완성하고 실행합니다
2. 아래 써 보기 칸을 실행합니다. 만든 함수를 불러 결과를 출력합니다. `# =>` 뒤가 나와야 할 결과입니다
3. 그 아래 확인 칸을 실행합니다. **아무것도 나오지 않으면 통과**입니다
4. 통과하지 못하면 `AssertionError`와 함께 무엇을 고쳐야 하는지 나옵니다

써 보기 칸과 확인 칸은 고치지 않고 실행만 합니다.

## 문제 1. 안전하게 바꾸기

문자열을 정수로 바꿔 돌려주는 함수를 완성합니다. 바꿀 수 없으면 `None`을 돌려줍니다. `try` · `except`로 씁니다.

```python
def safe_int(text: str) -> int | None:
    try:
        return int(text)
    except ValueError:
        return None
```

```python
# 써 보기
print(safe_int("42"), safe_int("-7"), safe_int("사십"))  # => 42 -7 None
```

```python
# 확인 칸
assert "safe_int" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert safe_int("42") == 42, "\"42\"는 42입니다"
assert safe_int("-7") == -7, "\"-7\"은 -7입니다. isdigit()은 -를 숫자로 보지 않습니다"
assert safe_int("사십") is None, "바꿀 수 없으면 None을 돌려줍니다. try · except ValueError"
```

## 문제 2. 나누기

`a / b`를 돌려주고, `b`가 0이면 `None`을 돌려주는 함수를 완성합니다. `if`로 0인지 검사하지 말고 `try`를 씁니다.

```python
def safe_divide(a: float, b: float) -> float | None:
    try:
        return a / b
    except ZeroDivisionError:
        return None
```

```python
# 써 보기
print(safe_divide(10, 4), safe_divide(1, 0))  # => 2.5 None
```

```python
# 확인 칸
assert "safe_divide" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert safe_divide(10, 4) == 2.5, "10 / 4 = 2.5"
assert safe_divide(10, 0) is None, "0으로 나누면 None입니다. except ZeroDivisionError"
```

## 문제 3. 점수 모으기

문자열 리스트를 받아, 정수로 바꿀 수 있는 것은 정수로 바꿔 한 리스트에, 바꾸지 못한 문자열은 다른 리스트에 모읍니다. 두 리스트를 `(성공, 실패)` 튜플로 돌려주는 함수를 완성합니다.

```python
def collect_scores(texts: list[str]):
    ok, bad = [], []
    for t in texts:
        try:
            ok.append(int(t))
        except ValueError:
            bad.append(t)
    return ok, bad
```

```python
# 써 보기
print(collect_scores(["80", "팔십", "95", "", "70"]))  # => ([80, 95, 70], ['팔십', ''])
```

```python
# 확인 칸
assert "collect_scores" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = collect_scores(["80", "팔십", "95", "", "70"])
assert type(result) == tuple, "두 리스트를 쉼표로 이어 함께 return 합니다 (return 성공, 실패)"
ok, bad = result
assert ok == [80, 95, 70], "바꿀 수 있는 것만 int()로 바꿔 순서대로 모읍니다"
assert bad == ["팔십", ""], "바꾸지 못한 문자열을 따로 모읍니다 (빈 문자열도 int()가 실패합니다)"
assert collect_scores([]) == ([], []), "빈 리스트면 ([], [])입니다"
```

## 문제 4. 잘못된 값 막기

나이(정수)를 받아 그대로 돌려주되, 0보다 작거나 150보다 크면 `ValueError`를 일으키는 함수를 완성합니다.

```python
def check_age(age: int) -> int:
    if age < 0 or age > 150:
        raise ValueError(f"나이가 범위를 벗어났습니다: {age}")
    return age
```

```python
# 써 보기
print(check_age(30))  # => 30
check_age(-1)  # => ValueError: 나이가 범위를 벗어났습니다: -1 (에러가 나면 맞습니다)
```

```python
# 확인 칸
assert "check_age" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert check_age(30) == 30, "범위 안의 나이는 그대로 돌려줍니다"
for good_age in (0, 150):
    try:
        check_age(good_age)
    except ValueError:
        assert False, f"check_age({good_age})는 범위 안입니다. 0보다 작거나 150보다 클 때만 일으킵니다"
for bad_age in (-1, 200):
    try:
        check_age(bad_age)
    except ValueError:
        pass
    else:
        assert False, f"check_age({bad_age})는 ValueError를 일으켜야 합니다. raise ValueError(...)"
```
