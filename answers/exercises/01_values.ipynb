# 01 연습

1. 빈 칸에 코드를 쓰고 실행합니다
2. 바로 아래 확인 칸을 실행합니다. **아무것도 나오지 않으면 통과**입니다
3. 통과하지 못하면 `AssertionError`와 함께 무엇을 고쳐야 하는지 나옵니다. `위 빈 칸을 먼저 채우고 실행해 주세요`가 나오면 빈 칸을 채워 실행했는지 확인합니다

확인 칸은 고치지 않고 실행만 합니다.

## 문제 1. 값의 종류

`price`에 정수 `1500`을, `rate`에 실수 `0.1`을, `item`에 문자열 `"커피"`를 넣습니다.

```python
price = 1500
rate = 0.1
item = "커피"
```

```python
# 확인 칸
assert "price" in globals() and "rate" in globals() and "item" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(price) == int and price == 1500, "price는 정수 1500입니다"
assert type(rate) == float and rate == 0.1, "rate는 실수 0.1입니다"
assert type(item) == str and item == "커피", "item은 문자열 \"커피\"입니다"
```

## 문제 2. 몫과 나머지

`seconds`는 `200`초입니다. 몇 분 몇 초인지 `m`(분)과 `s`(초)에 넣습니다. 숫자를 직접 쓰지 말고 `seconds`로 계산합니다.

```python
seconds = 200
m = seconds // 60
s = seconds % 60
```

```python
# 확인 칸
assert "m" in globals() and "s" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert m == 3, "200초는 3분하고 조금 더입니다. 몫은 //"
assert s == 20, "남은 초는 나머지 %로 구합니다"
```

## 문제 3. 형변환

`count_text`는 문자열 `"7"`입니다. 이 값을 정수로 바꿔 `3`을 더한 값을 `total`에 넣습니다.

```python
count_text = "7"
total = int(count_text) + 3
```

```python
# 확인 칸
assert "total" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert total != "73", "문자열끼리 이으면 \"73\"이 됩니다. int()로 바꿔서 더합니다"
assert total == 10, "int(count_text) + 3은 10입니다"
```

## 문제 4. f-string

`name`에 `"홍길동"`, `score`에 `90`을 넣고, f-string으로 `message`를 만듭니다. 결과는 `홍길동님의 점수는 90점입니다`여야 합니다.

```python
name = "홍길동"
score = 90
message = f"{name}님의 점수는 {score}점입니다"
```

```python
# 확인 칸
assert "message" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert message == f"{name}님의 점수는 {score}점입니다", "띄어쓰기와 글자를 확인해 주세요: 홍길동님의 점수는 90점입니다"
```

## 문제 5. 비교

`age`가 `19`일 때, 20 이상인지를 `is_adult`에 넣습니다. `True`나 `False`를 직접 쓰지 말고 비교 연산자로 만듭니다.

```python
age = 19
is_adult = age >= 20
```

```python
# 확인 칸
assert "is_adult" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(is_adult) == bool, "is_adult는 True나 False여야 합니다. 비교 연산자 >=를 씁니다"
assert is_adult == False, "19는 20 이상이 아닙니다"
```

## 문제 6. 복합 대입

`coin`에 `100`을 넣습니다. 그다음 **복합 대입 연산자만** 써서 `30`을 빼고, `2`를 곱합니다.

```python
coin = 100
coin -= 30
coin *= 2
```

```python
# 확인 칸
assert "coin" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert coin != 40, "곱하기를 먼저 했나요? 30을 뺀 다음에 2를 곱합니다"
assert coin == 140, "(100 - 30) × 2 = 140입니다"
```

## 문제 7. 계산 순서

`price`는 `1000`, `count`는 `3`, `discount`는 `500`입니다. **`price`를 `count`개 산 금액에서 `discount`를 뺀 값**에 다시 `2`를 곱해 `result`에 넣습니다. 한 줄로 쓰고, 필요한 곳에 괄호를 씁니다.

```python
price = 1000
count = 3
discount = 500
result = (price * count - discount) * 2
```

```python
# 확인 칸
assert "result" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert result != 2000, "괄호가 없으면 discount * 2를 먼저 계산합니다"
assert result == 5000, "(1000 × 3 - 500) × 2 = 5000입니다"
```
