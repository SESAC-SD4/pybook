## 문제 1. 인덱스

`fruits`에서 두 번째 과일을 `second`에, 마지막 과일을 `last`에 넣습니다. `last`는 음수 인덱스로 꺼냅니다.

```python
fruits = ["사과", "바나나", "포도", "딸기"]
second = fruits[1]
last = fruits[-1]
```

```python
# 확인 칸
assert "second" in globals() and "last" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert second != "사과", "인덱스는 0부터 시작합니다. 두 번째는 1번입니다"
assert second == "바나나", "두 번째 과일은 바나나입니다"
assert last == "딸기", "마지막 값은 [-1]로 꺼냅니다"
```

## 문제 2. 슬라이싱

`code`에서 슬라이싱으로 앞 세 글자를 `prefix`에, 뒤 두 글자(`PY`)를 `suffix`에 넣습니다.

```python
code = "SD4-2026-PY"
prefix = code[:3]
suffix = code[9:]
```

## 문제 3. 리스트 채우기

`numbers` 가운데 5보다 큰 수만 `big`에 담습니다. `big`을 빈 리스트로 시작해 `for`와 `append`로 채웁니다.

```python
numbers = [3, 8, 1, 9, 4]
big = []
# numbers의 수를 하나씩 확인해서
for n in numbers:
    # 5보다 크면
    if n > 5:
        # big에 담는다
        big.append(n)
```

```python
# 확인 칸
assert "big" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(big) == list, "big은 리스트여야 합니다"
assert big == [8, 9], "5보다 큰 수는 8과 9입니다. 순서도 numbers와 같아야 합니다"
```

## 문제 6. 딕셔너리 꺼내기

`book`의 `"price"` 값을 `book_price`에 넣습니다. `book`에 없는 키 `"author"`는 `.get()`으로 꺼내되, 없으면 `"미상"`이 되도록 `author`에 넣습니다.

## 문제 7. 딕셔너리 반복

`cart`는 메뉴별 개수, `price`는 메뉴별 가격입니다. `total`을 0에서 시작해 `cart.items()`로 반복하며 `개수 × 가격`을 더합니다.

```python
cart = {"라면": 2, "김밥": 3, "음료": 1}
price = {"라면": 4000, "김밥": 3000, "음료": 1500}
total = 0
for item, count in cart.items():
    total += count * price[item]
```

```python
# 확인 칸
assert "total" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert total == 18500, "2×4000 + 3×3000 + 1×1500 = 18500입니다"
```

## 문제 9. 나누고 잇기

`line`을 쉼표로 나눈 뒤 각 값의 앞뒤 빈칸을 지우고, `" / "`로 이어 `result`에 넣습니다. 결과는 `"python / sql / git"`입니다.

```python
line = "python, sql ,git"
# 풀어 쓴 모양
parts = []
for p in line.split(","):
    parts.append(p.strip())
result = " / ".join(parts)

# 컴프리헨션으로 줄인 모양
parts = [p.strip() for p in line.split(",")]
result = " / ".join(parts)
```

```python
# 확인 칸
assert "result" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(result) == str, "result는 문자열이어야 합니다. join을 씁니다"
assert result == "python / sql / git", "빈칸이 남았는지 확인해 주세요. 값마다 strip()을 씁니다"
```

## 문제 10. 컴프리헨션

`temps`에서 25 이상인 온도만 골라 담은 리스트를 **컴프리헨션 한 줄로** `hot`에 넣습니다.

```python
temps = [18, 25, 31, 22, 28]
# 풀어 쓴 모양
hot = []
for t in temps:
    if t >= 25:
        hot.append(t)

# 컴프리헨션으로 줄인 모양
hot = [t for t in temps if t >= 25]
```

```python
# 확인 칸
assert "hot" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert hot == [25, 31, 28], "25 이상이면 25도 들어갑니다. if t >= 25"
```

## 문제 11. 짝지어 딕셔너리 만들기

`keys`와 `values`를 `zip`으로 같은 자리끼리 짝지어 딕셔너리 `profile`을 만듭니다.

```python
keys = ["name", "level"]
values = ["홍길동", 3]
profile = dict(zip(keys, values))
```

```python
# 확인 칸
assert "profile" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert type(profile) == dict, "profile은 딕셔너리여야 합니다. dict(zip(keys, values))"
assert profile == {"name": "홍길동", "level": 3}, "keys와 values를 같은 자리끼리 짝짓습니다"
```
