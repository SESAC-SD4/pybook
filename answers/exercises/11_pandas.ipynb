```python
# 준비: 이 칸을 먼저 실행합니다 (수업과 같은 타이타닉 표 df)
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
df.shape
```

## 문제 1. 항구마다 몇 명

표 `df`와 칸 이름을 받아, 그 칸의 **값마다 몇 명인지** 돌려주는 함수를 완성합니다.

```python
def count_by(df, column: str):
    return df[column].value_counts()
```

```python
# 써 보기
print(count_by(df, "embark_town"))  # => Southampton 644 · Cherbourg 168 · Queenstown 77 (값마다 한 줄, 맨 아래 Name: count)
```

```python
# 확인 칸
assert "count_by" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = count_by(df, "embark_town")
assert result is not None, "print가 아니라 return으로 돌려줍니다"
assert result["Southampton"] == 644, "Southampton에서 탄 사람은 644명입니다. df[column].value_counts()"
assert count_by(df, "sex")["female"] == 314, "column 자리의 칸을 셉니다. \"embark_town\"을 직접 쓰지 않습니다"
```

## 문제 2. 혼자 탄 사람의 생존율

표 `df`를 받아, **혼자 탄 사람(`alone`이 `True`)의 생존율**을 돌려주는 함수를 완성합니다.

```python
def alone_rate(df) -> float:
    return df[df["alone"] == True]["survived"].mean()
```

```python
# 써 보기
print(round(alone_rate(df), 3))  # => 0.304
```

```python
# 확인 칸
assert "alone_rate" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = alone_rate(df)
assert result is not None, "print가 아니라 return으로 돌려줍니다"
assert round(result, 3) == 0.304, "혼자 탄 줄을 고른 뒤 [\"survived\"].mean()입니다. 전체 생존율(0.384)이 나왔다면 줄을 고르지 않은 것입니다"
```

## 문제 3. 부자는 살았을까?

표 `df`와 숫자 `n`을 받아, **요금(`fare`)이 가장 비싼 `n`명의 생존율**을 돌려주는 함수를 완성합니다.

```python
def rich_rate(df, n: int) -> float:
    top = df.sort_values("fare", ascending=False).head(n)
    return top["survived"].mean()
```

```python
# 써 보기
print(rich_rate(df, 50), round(df["survived"].mean(), 3))  # => 0.76 0.384 (비싼 50명과 전체. 0.74도 맞습니다)
```

```python
# 확인 칸
assert "rich_rate" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = rich_rate(df, 50)
assert result is not None, "print가 아니라 return으로 돌려줍니다"
# 50번째 자리에 요금이 같은 두 명(생존 · 사망)이 있어 누가 들어가느냐에 따라 0.76 또는 0.74
assert result in (0.76, 0.74), "fare가 큰 순서로 정렬해(ascending=False) 앞 n명의 [\"survived\"].mean()입니다"
assert rich_rate(df, 10) in (0.7, 0.8), "n 자리의 숫자만큼 고릅니다. head(n)"
```

## 문제 4. 산 사람과 아닌 사람의 나이

표 `df`를 받아, **`survived` 값마다 평균 나이**를 돌려주는 함수를 완성합니다.

```python
def age_by_survived(df):
    return df.groupby("survived")["age"].mean()
```

```python
# 써 보기
print(age_by_survived(df).round(1))  # => 0 30.6 · 1 28.3 (값마다 한 줄, 맨 아래 Name: age)
```

```python
# 확인 칸
assert "age_by_survived" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = age_by_survived(df)
assert result is not None, "groupby 결과를 return합니다"
assert round(result[1], 1) == 28.3, "산 사람(1)의 평균 나이는 28.3입니다. groupby(\"survived\")[\"age\"].mean()"
```
