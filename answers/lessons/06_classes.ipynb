## 4. 메서드

**기본 구조**

```
class 클래스이름:
    def 메서드이름(self, 매개변수):
        실행할 코드 (self.속성을 쓴다)

인스턴스.메서드이름(인수)
```

메서드의 첫 매개변수는 늘 `self`이고, `self`로 그 인스턴스의 속성을 씁니다.

````python
#> Student에 평균을 돌려주는 average(self) 메서드를 더해 다시 만듭니다
class Student():

    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def average(self):
        return sum(self.scores) / len(self.scores)

    def hi(self):
        return 'hi'
````

````python
#> s1 · s2를 다시 만들고, 이름과 평균을 한 줄씩 출력합니다
s1 = Student('민수', [90, 80, 100])
s2 = Student('지은', [75, 100, 50])
s1.average()
s2.average()
s1.hi()
````

````python
#> x와 y를 받는 Point 클래스를 만들고, __str__으로 "(x, y)" 모양 문자열을 돌려줍니다
class Point():
    def __init__(self, n, m):
        self.x = n
        self.y = m
    
    def __str__(self):
        return f'{self.x}, {self.y}'

a = Point(3, 4)
b = Point(1, 2)
c = Point(-5, 3)

print(a)
print(a.y)
````

두 점 사이의 거리는 `math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)`입니다.

````python
#> math를 가져와, 다른 Point를 받아 두 점 사이의 거리를 돌려주는 distance(self, other)를 더해 Point를 다시 만듭니다
import math

class Point():
    def __init__(self, n, m):
        self.x = n
        self.y = m
    
    def __str__(self):
        return f'{self.x}, {self.y}'

    def distance(self, other):
        x1 = self.x
        x2 = other.x

        y1 = self.y
        y2 = other.y
        
        return math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
````

````python
#> Point(0, 0)을 a에, Point(3, 4)를 b에 넣고, a에서 b까지의 거리를 출력합니다
a = Point(0, 0)
b = Point(3, 4)

a.distance(b)
````

`a.distance(b)`를 부르면 `self` 자리에 `a`, `other` 자리에 `b`가 들어갑니다.

````python
#> 중심 center(Point 인스턴스)와 반지름 r을 받는 Circle 클래스를 만듭니다
class Circle():
    def __init__(self, center, r):
        self.center = center
        self.r = r
    

c = Circle(a, 4)
print(c.center)

c2 = Circle(Point(3, 3), 5)
print(c2.center)
````

````python
#> Circle(Point(0, 0), 5)로 c를 만들고, c.center와 c.center.x와 c.r을 print 한 번으로 출력합니다
print(c2.center, c2.center.x, c2.r)
````

`c.center`는 `Point` 인스턴스입니다. 점을 한 번 더 찍으면(`c.center.x`) 그 인스턴스의 속성을 씁니다.

````python
#> 점 하나를 받아 원 안(경계 포함)에 있으면 True를 돌려주는 contains(self, point)를 더해 Circle을 다시 만듭니다
class Circle():
    def __init__(self, center, r):
        self.center = center
        self.r = r

    def contains(self, point):
        return self.center.distance(point) <= self.r
        
````

````python
#> Circle(Point(0, 0), 5)로 c를 다시 만들고, Point(3, 4)와 Point(6, 0)이 원 안에 있는지 print 한 번으로 출력합니다
c = Circle(Point(0, 0), 5)
c.contains(Point(6, 4))
````

`contains`는 거리 계산을 중심점의 `distance`에 맡깁니다(`self.center.distance(point)`).

## 8. `@classmethod`

**기본 구조**

```
class 클래스이름:
    @classmethod
    def 메서드이름(cls, 매개변수):
        return cls(인수)

클래스이름.메서드이름(인수)
```

`@classmethod`를 붙인 메서드는 인스턴스가 아니라 **클래스**를 첫 인수(`cls`)로 받습니다. 다른 모양의 값으로 인스턴스를 만드는 메서드에 자주 씁니다.

````python
#> "민수,80,90,70" 같은 문자열로 Student를 만드는 from_text(cls, text) 클래스 메서드를 가진 Student를 만듭니다
class Student():
    # 클래스 변수

    # 인스턴스 매서드
    def __init__(self, name, scores):
        # 인스턴스변수
        self.name = name
        self.scores = scores

    # 클래스 매서드
    @classmethod
    def from_text(cls, text):
        # text => 민수,80,90,70
        parts = text.split(',')

        # parts => [민수, 80, 90, 70]
        
        # return Student(parts[0], [int(p) for p in parts[1:]])
        return cls(parts[0], [int(p) for p in parts[1:]])
        
````

````python
#> Student.from_text("민수,80,90,70")으로 s를 만들고, 이름과 점수를 print 한 번으로 출력합니다
s1 = Student('민수', [80, 90, 60])
s3 = Student.from_text('민수,80,90,60')
print(s3.name, s3.scores)

s4 = s3.from_text('길동,40,20,30')
print(s4.name, s4.scores)

````

- `@`로 시작하는 줄은 바로 아래 함수에 기능을 덧붙입니다. 이것을 **데코레이터**라고 하고, 09장에서 직접 만듭니다
- `parts[1:]`는 0번을 뺀 나머지입니다(03장 슬라이싱)

## 9. 상속

**기본 구조**

```
class 부모:
    def __init__(self, 매개변수):
        self.속성 = 매개변수

class 자식(부모):
    def __init__(self, 매개변수, 매개변수2):
        super().__init__(매개변수)
        self.속성2 = 매개변수2
```

이미 있는 클래스를 바탕으로 새 클래스를 만들 수 있습니다. 이것을 **상속**이라고 합니다. 바탕이 되는 클래스를 **부모**, 새 클래스를 **자식**이라고 부릅니다.

````python
#> name을 받고, introduce()가 "저는 OO입니다"를 돌려주는 Person 클래스를 만듭니다
class Person():
    def __init__(self, name, phone, birth, address):
        self.name = name
        self.phone = phone
        self.birth = birth
        self.address = address

    def introduce(self):
        return f'저는 {self.name}입니다.'
    
````

````python
#> Person을 상속한 Student를 만듭니다. __init__에서 super().__init__(name)을 부르고 school을 더합니다
class Student(Person):
    def __init__(self, name, phone, birth, address, student_id):
        super().__init__(name, phone, birth, address)
        # self.name = name
        # self.phone = phone
        # self.birth = birth
        # self.address = address
        self.student_id = student_id

s1 = Student('민수', 123456)
s1.introduce()
````

- `super().__init__(name)` — 부모의 `__init__`을 불러 `self.name`을 넣습니다

````python
#> Student("민수", "SeSAC")으로 s를 만들고, s.introduce()와 s.school을 print 한 번으로 출력합니다
````

`Student`에는 `introduce`를 만들지 않았지만 부모의 것을 그대로 씁니다.

자식에서 같은 이름의 메서드를 다시 만들면 자식 것이 쓰입니다(**오버라이딩**).

````python
#> introduce()를 "저는 SeSAC 학생 OO입니다" 모양으로 다시 만든 Student를 만들고, 불러 봅니다
class Soldier(Person):
    def introduce(self):
        return '충성'
        
s1 = Soldier('대한')
s1.introduce()

s2 = Student('민국', 12354)
s2.introduce()

    
````

`isinstance(값, 클래스)`는 값이 그 클래스(또는 그 자식 클래스)의 인스턴스인지 `True` · `False`로 알려 줍니다.

## 10. 추상화

**기본 구조**

```
from abc import ABC, abstractmethod

class 부모(ABC):
    @abstractmethod
    def 메서드이름(self):
        pass

class 자식(부모):
    def 메서드이름(self):
        실제 코드
```

도형마다 넓이를 구하는 법은 다르지만, "넓이를 구할 수 있다"는 약속은 같습니다. **추상 클래스**는 그 약속만 정하고, 실제 구현은 자식에게 맡깁니다.

````python
#> abc 모듈에서 ABC와 abstractmethod를 가져와, area()를 추상 메서드로 가진 Shape 클래스를 만듭니다
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
````

````python
#> Shape()로 인스턴스를 만들어 봅니다
s = Shape()
````

추상 클래스는 직접 인스턴스를 만들 수 없습니다. 자식이 추상 메서드를 모두 만들어야 인스턴스를 만들 수 있습니다.

````python
#> Shape를 상속한 Rectangle(width, height)과 Triangle(base, height)을 만들고 각각 area()를 구현합니다

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return self.base * self.height / 2
    

        
````

````python
#> shapes 리스트에 Rectangle(3, 4)와 Triangle(3, 4)를 담고, for로 각 도형의 넓이를 출력합니다
shapes = [Rectangle(3, 4), Triangle(3, 4)]
for shape in shapes:
    print(shape.area())
````

`for` 안의 코드는 도형이 무엇인지 몰라도 `area()`만 부릅니다. 같은 이름의 메서드를 종류마다 다르게 동작하게 하는 것을 **다형성**이라고 합니다.

## 11. `dataclass`

**기본 구조**

```
from dataclasses import dataclass

@dataclass
class 클래스이름:
    속성: 종류
    속성: 종류 = 기본값
```

값을 담는 일이 주된 클래스는 `__init__`과 출력 모양을 매번 쓰기 번거롭습니다. `@dataclass`를 붙이고 **타입힌트(04장)** 로 속성을 적으면 그것들을 만들어 줍니다.

````python
#> dataclasses에서 dataclass를 가져와, name: str과 scores: list[int]를 가진 Student를 @dataclass로 만듭니다
from dataclasses import dataclass

@dataclass
class Student():
    name: str
    scores: list[int]
    school: str = 'sesac'

    def average(self):
        return sum(self.scores) / len(self.scores)

````

````python
#> Student("민수", [80, 90, 70])으로 s를 만들고, s와 s.average()를 print 한 번으로 출력합니다
s = Student('민수', [80, 90, 70])
print(s, s.average())
````

````python
#> 같은 값으로 만든 두 Student를 ==로 비교합니다
s1 = Student('민수', [80, 90, 70])
s2 = Student('민수', [80, 90, 70])
print(s1 == s2, s1 is s2)
````

`=`로 기본값을 주면 만들 때 빼도 됩니다. 기본값이 있는 속성은 없는 속성 뒤에 씁니다.

````python
#> school: str = "SeSAC"을 더한 Student를 만들고, Student("민수", [80])을 출력합니다
````

`@dataclass`는 `__init__` · 보기 좋은 출력 · `==` 비교를 만들어 줍니다. 4차시 FastAPI에서 쓰는 Pydantic 모델도 이렇게 타입힌트로 속성을 적는 클래스입니다.

## 정리

| 할 일 | 방법 |
| --- | --- |
| 틀 만들기 · 인스턴스 만들기 | `class 이름:` · `이름(…)` |
| 처음 값 넣기 | `def __init__(self, …):` · `self.속성 = 값` |
| 메서드 | 첫 매개변수 `self` · `인스턴스.메서드()` |
| `print` 모양 | `def __str__(self):` |
| 객체를 담기 · 주고받기 | `self.속성 = 다른인스턴스` · `def 메서드(self, other):` |
| 모두 같이 쓰는 값 | 클래스 변수 — `클래스.이름`으로 바꾼다 |
| 다른 모양으로 만들기 | `@classmethod` · `cls(…)` |
| 물려받기 | `class 자식(부모):` · `super().__init__(…)` · 오버라이딩 |
| 약속만 정하기 | `ABC` · `@abstractmethod` |
| 값을 담는 클래스 | `@dataclass` + 타입힌트 |

| 에러 | 뜻 |
| --- | --- |
| `AttributeError` | 그런 속성 · 메서드가 없음 |
| `TypeError: … missing … required positional argument` | 만들 때 · 부를 때 인수를 빼먹음 |
| `TypeError: … takes 0 positional arguments but 1 was given` | 메서드에 `self`를 빼먹음 |
| `TypeError: Can't instantiate abstract class …` | 추상 클래스는 직접 만들 수 없음 |
