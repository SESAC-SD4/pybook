## 1. 멈추는 프로그램

````python
texts = ["80", "95", "팔십", "70"]

total = 0
for t in texts:
    total += int(t)
    print(t, "더함")
print("합계", total)
````

`"팔십"`에서 `ValueError`가 나고, 그 뒤의 `"70"`과 `합계`는 실행되지 않았습니다. 실행 중에 나는 에러를 **예외(exception)** 라고 합니다.

## 2. 에러 읽기

에러가 함수 안에서 나면, 칸 아래에 **어떤 함수가 어떤 함수를 불렀는지**가 순서대로 나옵니다. 이것을 **트레이스백(traceback)** 이라고 합니다.

````python
#> 문자열을 정수로 바꾸는 to_int(text)와, 리스트의 합을 to_int로 구하는 total(texts)를 만들고 total(["1", "둘"])을 불러 봅니다

def to_int(text):
    return int(text)

def total(texts):
    result = 0
    for t in texts:
        result += to_int(t)
    return result

total(['10', '20', '삼십'])

````

트레이스백은 아래에서 위로 읽습니다.

| 읽는 순서 | 무엇을 보나 |
| --- | --- |
| 1 | **마지막 줄** — 에러 종류와 내용 (`ValueError: invalid literal …`) |
| 2 | 그 바로 위 — 에러가 **난** 줄 (`to_int` 안의 `int(text)`) |
| 3 | 더 위 — 그 함수를 **부른** 줄 (`total` 안 → 칸의 `total([…])`) |

## 3. `try` · `except`

**기본 구조**

```
try:
    에러가 날 수 있는 코드
except 에러종류:
    그 에러가 났을 때 할 일
```

````python
int('팔십')
````

````python
#> try 안에서 int("팔십")을 하고, ValueError가 나면 "숫자가 아닙니다"를 출력합니다
try:
    int('팔십')
except ValueError:
    print('숫자가 아닙니다.')
````

1절의 반복을 고쳐, 숫자가 아닌 값은 건너뛰고 끝까지 갑니다.

````python
#> texts의 값을 하나씩 int로 더해 "더함"을 출력하되, ValueError가 나면 "건너뜀"을 출력하고 다음 값으로 갑니다. 끝나고 합계를 출력합니다
texts = ['20', '30', '팔십', '100']
total = 0

for text in texts:
    try:
        total += int(text)
    except ValueError:
        print(f'{text} 건너뜀')
print(total)
````

`try` 안에서 에러가 나면 그 줄 아래의 `try` 코드는 실행하지 않고 바로 `except`로 갑니다.

## 4. 여러 에러 · 에러 메시지 받기

**기본 구조**

```
try:
    에러가 날 수 있는 코드
except 에러종류1 as e:
    할 일
except (에러종류2, 에러종류3) as e:
    할 일
```

`as e`를 붙이면 에러를 이름 `e`로 받습니다. `print(e)`로 메시지를 봅니다.

````python
#> a = 10, b = "0"일 때 a / int(b)를 계산합니다. ZeroDivisionError와 ValueError를 각각 except로 받아 메시지를 출력합니다

a = 10
b = '0'

try:
    a / int(b)
except ZeroDivisionError as e:
    print('0으로 나눌수 없습니다.', e)
except ValueError as e:
    print('숫자가 아닙니다.', e)

````

`b`를 `"둘"`로 바꿔 다시 실행해 봅니다.

여러 에러를 한 `except`로 받으려면 괄호로 묶습니다.

````python
#> (ZeroDivisionError, ValueError)를 한 except로 받아, 에러의 type과 메시지를 출력합니다 (b는 "둘")
a = 1
b = 'ㅁㄴㅇㄹ'
try:
    a / int(b)
except (ZeroDivisionError, ValueError) as e:
    print(e)
````

| 에러 | 언제 |
| --- | --- |
| `ValueError` | 종류는 맞는데 값이 맞지 않음 (`int("팔십")`) |
| `TypeError` | 종류가 맞지 않음 (`"나이" + 25`) |
| `ZeroDivisionError` | 0으로 나눔 |
| `KeyError` · `IndexError` | 없는 키 · 범위를 벗어난 번호 |
| `FileNotFoundError` | 없는 파일을 엶 (08장) |

## 5. `else` · `finally`

**기본 구조**

```
try:
    에러가 날 수 있는 코드
except 에러종류:
    에러가 났을 때
else:
    에러가 나지 않았을 때
finally:
    에러가 나든 안 나든 늘
```

````python
#> 문자열을 받아 int로 바꾸는 함수 convert(text)를 만듭니다. try · except ValueError · else · finally에서 각각 다른 말을 출력합니다. "7"과 "칠"로 불러 봅니다

def convert(text):
    try:
        n = int(text)
    except ValueError:
        print('실패: ', text)
    else:
        print('성공: ', n)
    finally:
        print('끝')
    
convert('7')
convert('칠')
    
````

`finally`는 파일을 닫거나 연결을 끊는 것처럼 **무슨 일이 있어도 해야 하는 정리**에 씁니다. 08장에서는 `with`가 이 일을 대신합니다.

## 6. 넓게 잡기

`except Exception:`은 거의 모든 에러를 잡습니다. 편해 보이지만, **내가 생각하지 못한 에러까지** 숨깁니다.

````python
#> try 안에서 이름을 잘못 쓴 print(totl)을 하고, except Exception으로 잡아 "숫자가 아닙니다"를 출력해 봅니다
try:
    print(asdf)
except Exception as e:
    print(e)
````

이름을 잘못 쓴 `NameError`인데 "숫자가 아닙니다"가 나왔습니다. 에러를 잡을 때는 **예상하는 에러만** 적습니다. `except:`처럼 종류를 아예 적지 않는 모양도 쓰지 않습니다.

`ValueError` · `TypeError` 같은 에러는 모두 `Exception`의 자식입니다(06장 상속). 그래서 `except Exception:`이 거의 모든 에러를 잡습니다.

## 7. `raise`

**기본 구조**

```
if 잘못된 조건:
    raise 에러종류("메시지")
```

잘못된 값이 들어왔을 때 **내 코드가 직접 에러를 일으킬** 수 있습니다. 조용히 넘어가지 않고 부른 쪽에 알리는 방법입니다.

````python
#> balance와 amount를 받아 출금 후 잔액을 돌려주는 withdraw를 만듭니다. amount가 0 이하이거나 balance보다 크면 ValueError를 raise합니다
def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError('출금액은 0원보다는 커야합니다.')
    if amount > balance:
        raise ValueError('잔액이 부족합니다.')
    return balance - amount
````

````python
#> withdraw(10000, -500)을 불러 봅니다
withdraw(10000, 20000)
````

````python
#> withdraw(10000, 50000)을 try로 감싸 부르고, ValueError 메시지를 출력합니다
try:
    withdraw(10000, 50000)
except ValueError:
    print('출금실패')
````

## 8. 미리 검사하기와 `try`

같은 일을 두 방법으로 할 수 있습니다.

`문자열.isdigit()`는 모든 글자가 숫자이면 `True`를 돌려줍니다.

````python
#> text = "123"일 때, if와 isdigit()으로 숫자인지 먼저 검사해 int로 바꾸는 코드를 씁니다
text = '123'
if text.isdigit():
    print(int(text))
else:
    print('숫자아님')
````

````python
#> 같은 일을 try · except ValueError로 씁니다
text = '일이삼'
try:
    print(int(text))
except ValueError:
    print('숫자아님')
````

`text`를 `"-5"`로 바꿔 두 칸을 다시 실행해 봅니다.

`isdigit()`은 `-` 기호를 숫자로 보지 않아 `"-5"`를 "숫자 아님"으로 봅니다. `int()`가 받아들이는 모양을 미리 모두 검사하기는 어렵습니다. 바꿔 보고 실패하면 처리하는 `try` 쪽이 Python에서 흔한 방법입니다.

## 정리

| 할 일 | 방법 |
| --- | --- |
| 에러 읽기 | 마지막 줄 → 난 줄 → 부른 줄 (아래에서 위로) |
| 에러가 나면 다른 일 하기 | `try:` · `except 에러종류:` |
| 메시지 받기 | `except 에러종류 as e:` · 여러 종류는 `(A, B)` |
| 에러가 없을 때 · 늘 | `else:` · `finally:` |
| 잡지 말 것 | 예상하는 에러만 적는다 (`except Exception:` · `except:` 쓰지 않음) |
| 직접 일으키기 | `raise ValueError("메시지")` |
