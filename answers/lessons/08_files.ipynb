## 2. 파일 쓰기

**기본 구조**

```
f = open("파일이름", "w", encoding="utf-8")
f.write(글)
f.close()
```

````python
#> memo.txt를 쓰기 모드("w")로 열어 "첫 메모"를 쓰고 닫습니다
f = open('memo.txt', 'w', encoding='utf-8')
f.write('첫 메모입니다.')
f.close()
````

왼쪽 파일 목록에 `memo.txt`가 생겼습니다. 두 번 클릭해 열어 봅니다.

- `"w"`는 쓰기 모드입니다. **같은 이름의 파일이 있으면 내용을 지우고** 새로 씁니다
- `encoding="utf-8"`은 한글을 저장하는 방식입니다. 늘 붙입니다(윈도우와 맥에서 같은 파일이 되게)

## 3. `with`

**기본 구조**

```
with open("파일이름", "모드", encoding="utf-8") as f:
    f.write(글)            # 블록이 끝나면 저절로 닫힌다
```

`close()`를 빼먹으면 내용이 저장되지 않을 수 있습니다. `with`로 열면 블록이 끝날 때 **저절로 닫습니다**(07장 `finally`가 하던 정리).

````python
#> with로 memo.txt를 쓰기 모드로 열어, "첫째 줄"과 "둘째 줄"을 \n으로 나눠 씁니다
with open('memo.txt', 'w', encoding='utf-8') as f:
    f.write('추가 텍스트\n')
    f.write('두번째 텍스트\n')
    f.write('세번째 텍스트\n')
    f.write('네번째 텍스트\n')
````

`.write()`는 줄을 바꿔 주지 않습니다. 줄을 바꾸려면 `\n`을 씁니다(01장 1절).

## 4. 파일 읽기

**기본 구조**

```
with open("파일이름", encoding="utf-8") as f:
    전체 = f.read()

with open("파일이름", encoding="utf-8") as f:
    for 줄 in f:
        줄.strip()
```

````python
#> with로 memo.txt를 읽기 모드("r")로 열어 .read()로 전체를 읽고 확인합니다
with open('memo.txt', 'r', encoding='utf-8') as f:
    text = f.read()
    
    
````

모드를 적지 않으면 읽기(`"r"`)입니다. `for`로 한 줄씩 꺼낼 수도 있습니다. 줄 끝의 `\n`은 `.strip()`으로 지웁니다(03장 8절).

````python
#> with로 memo.txt를 열어 for로 한 줄씩 꺼내, strip한 줄을 출력합니다
with open('memo.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line.strip())
````

| 모드 | 뜻 | 파일이 있으면 | 없으면 |
| --- | --- | --- | --- |
| `"r"` | 읽기 | 읽는다 | `FileNotFoundError` |
| `"w"` | 쓰기 | **지우고** 새로 쓴다 | 만든다 |
| `"a"` | 덧붙이기 | 끝에 이어 쓴다 | 만든다 |

### 덧붙이기

```
with open("파일이름", "a", encoding="utf-8") as f:

```

````python
#> memo.txt를 "a" 모드로 열어 "셋째 줄"을 덧붙이고, 다시 읽어 확인합니다
with open('memo.txt', 'a', encoding='utf-8') as f:
    f.write('기존데이터 그대로 뒤에 붙는 내용')
````

## 6. 경로

**기본 구조**

```
from pathlib import Path

경로 = Path("폴더") / "파일이름"
Path("폴더").mkdir(exist_ok=True)
r"C:\폴더\파일"                  # 윈도우 경로는 r을 붙인다
```

파일 이름만 쓰면 **이 노트북이 있는 폴더**에서 찾습니다. 다른 폴더의 파일은 경로로 가리킵니다.

`pathlib`의 `Path`를 쓰면 경로를 `/`로 이어 붙일 수 있고, 윈도우와 맥에서 같게 동작합니다.

````python
#> pathlib에서 Path를 가져와, data 폴더를 만들고(exist_ok=True) 그 안의 notes.txt 경로를 path에 넣어 확인합니다
from pathlib import Path

folder = Path('data')
folder.mkdir(exist_ok=True)
path = folder / 'notes.txt' # => data/notes.txt
print(path)
````

`exist_ok=True`는 폴더가 이미 있어도 에러를 내지 않게 합니다. `path`는 `PosixPath`(윈도우는 `WindowsPath`)로 보이는 경로 값이고, `open()`에 그대로 넣을 수 있습니다.

````python
#> path에 "폴더 안의 파일"을 쓰고, path.exists()와 path.read_text(encoding="utf-8")를 print 한 번으로 출력합니다
with open(path, 'w') as f:
    f.write('hello world')

path2 = Path('asdf') 

print(path.exists(), path2.exists(), path.read_text())
    
````

`path.exists()`는 파일이 있으면 `True`, `path.read_text()`는 열기 · 읽기 · 닫기를 한 번에 해서 내용을 돌려줍니다.

윈도우 경로는 `\`로 이어지는데, 문자열 안의 `\`는 이스케이프 문자입니다(01장 1절). 그대로 쓰려면 따옴표 앞에 `r`을 붙입니다(**raw string**).

````python
#> "C:\new_folder"와 r"C:\new_folder"를 각각 print합니다
print('C:\new_folder')
print(r"C:\new_folder")
````

`\n`이 줄 바꿈으로 읽혀 경로가 깨졌습니다. 경로는 `r"…"`이나 `Path`와 `/`를 씁니다.

## 7. CSV

**기본 구조**

```
import csv

with open("파일.csv", "w", encoding="utf-8", newline="") as f:
    csv.writer(f).writerows(줄들)

with open("파일.csv", encoding="utf-8", newline="") as f:
    for 줄 in csv.reader(f):             # 줄은 문자열 리스트
        실행할 코드
```

**CSV**는 쉼표로 칸을 나눈 표 파일입니다. 구글 시트 · 엑셀에서 표로 열립니다(엑셀에서 한글이 깨지면 구글 시트로 엽니다). `csv` 모듈로 씁니다.

````python
#> csv를 가져와, rows를 scores.csv에 writer로 씁니다 (newline="")
rows = [["이름", "점수"], ["민수", 80], ["지은", 95]]

import csv

with open('scores.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f) 
    writer.writerows(rows)
````

`newline=""`은 CSV를 쓸 때 줄이 두 번 바뀌는 것을 막습니다. CSV를 열 때는 늘 붙입니다.

````python
#> scores.csv를 csv.reader로 읽어 한 줄씩 출력합니다
with open('scores.csv', 'r', encoding='utf-8', newline='') as f:
    for row in csv.reader(f):
        print(row)
        
````

읽은 값은 **모두 문자열**입니다(`'80'`). 계산하려면 `int()`로 바꿉니다. 표 모양 데이터는 12장 pandas로 더 쉽게 읽고 씁니다.

## 8. JSON

**기본 구조**

```
import json

with open("파일.json", "w", encoding="utf-8") as f:
    json.dump(값, f, ensure_ascii=False, indent=2)

with open("파일.json", encoding="utf-8") as f:
    값 = json.load(f)
```

**JSON**은 딕셔너리 · 리스트 모양 그대로 저장하는 글 형식입니다. 1차시 텔레그램 봇이 받은 메시지, 10장에서 받을 Open API 응답이 JSON입니다.

````python
#> json을 가져와, user를 user.json에 json.dump로 씁니다 (ensure_ascii=False, indent=2)
user = {
    "name": "민수", 
    "age": 25, 
    "skills": ["python", "sql"], 
    "is_member": True, 
    "memo": None
}

import json

with open('user.json', 'w', encoding='utf-8') as f:
    json.dump(user, f, ensure_ascii=False, indent=2)
    
````

`user.json`을 파일 목록에서 열어 봅니다. `True`가 `true`, `None`이 `null`로 바뀌어 있습니다.

| Python | JSON |
| --- | --- |
| `dict` | `{ }` (객체) |
| `list` | `[ ]` (배열) |
| `str` | `"…"` |
| `int` · `float` | 숫자 |
| `True` · `False` | `true` · `false` |
| `None` | `null` |

- `ensure_ascii=False` — 한글을 `\ubbfc` 같은 코드 대신 그대로 씁니다
- `indent=2` — 사람이 읽기 좋게 줄을 나눕니다

````python
#> user.json을 json.load로 읽어 loaded에 넣고, 종류와 loaded["skills"]를 print 한 번으로 출력합니다
with open('user.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)
    print(loaded['skills'])
````

파일 대신 문자열로 바꿀 때는 `json.dumps` · `json.loads`(끝에 `s`)를 씁니다.

````python
#> json.dumps로 user를 문자열 text로 바꾸고, 종류와 앞 30글자를 print 한 번으로 출력합니다
text = json.dumps(user)
print(text, type(text), text[:30])
````

## 정리

| 할 일 | 방법 |
| --- | --- |

| 모드 | `"r"` 읽기 · `"w"` 지우고 쓰기 · `"a"` 덧붙이기 |
