## 1. `if`와 들여쓰기

````python
#> age에 25를 넣고, age가 20 이상이면 "성인입니다"를 출력합니다
age = 10

if age >= 20:
    print('성인입니다.')
    
print('안녕하세요')
````

- `if 조건:` — 조건이 `True`면 아래 줄을 실행합니다. 끝에 **콜론(`:`)** 을 붙입니다
- 실행할 줄은 **들여쓰기(스페이스 4칸)** 합니다. 들여 쓴 줄까지가 `if`에 속합니다

`age`를 `15`로 바꾸면 어떻게 될까요?

````python
#> 같은 코드에서 age만 15로 바꿔 실행합니다
age = 15

if age >= 20:
    print('성인입니다.')
````

들여쓰기를 빼면 어떻게 될까요?

````python
#> 위 코드에서 print 줄의 들여쓰기를 빼고 실행해 봅니다
if age >= 20:
print('성인입니다.')
````

`IndentationError`는 **들여쓰기 에러**입니다. Python은 들여쓰기로 어디까지가 `if` 안인지 판단합니다.

### `pass`

`if` 아래에 아직 쓸 코드가 없을 때도 들여 쓴 줄이 하나는 있어야 합니다. 그 자리에 **아무것도 하지 않는** `pass`를 둡니다.

````python
#> age가 20 이상일 때 할 일을 아직 정하지 않았다고 보고, if 아래에 pass를 둡니다
if age >= 20:
    pass
````

## 2. `else`와 `elif`

`else:`는 조건이 `False`일 때 실행합니다.

````python
#> age가 20 이상이면 "성인입니다", 아니면 "미성년자입니다"를 출력합니다 (age는 15)
age = 15
if age >= 20:
    print('성인입니다')
else:
    print('미성년자입니다.')
````

갈래가 셋 이상이면 `elif 조건:`을 씁니다. **위에서부터 차례로 검사해서 처음 맞는 하나만** 실행합니다.

1차시 텔레그램 봇이 명령을 고르던 코드와 같은 모양입니다.

````python
#> text가 "/start"면 "안녕하세요!", "/lunch"면 "오늘은 라면", 그 밖에는 "모르는 명령입니다"를 output에 넣고 확인합니다 (text는 "/lunch")

text = '/lotto'

if text == '/start':
    output = '안녕하세요!'
elif text == '/lunch':
    output = '오늘은 라면'
elif text == '/lotto':
    output = '10, 20'
else:
    output = '모르는 명령어입니다.'

print(output)
````

`text`를 `"/start"`, `"/hello"`로 바꿔 다시 실행해 봅니다.

점수로 학점을 매겨 봅니다. 조건의 **순서**가 중요합니다.

````python
#> score가 90 이상이면 "A", 80 이상이면 "B", 그 밖에는 "C"를 grade에 넣고 확인합니다 (score는 85)

score = 85

if score >= 90:
    print('A')
elif score >= 80:
    print('B')
else:
    print('C')
````

## 4. `for`와 `range()`

`range(5)`는 `0`부터 `4`까지 다섯 개의 수를 만듭니다. `for`는 그 수를 하나씩 이름에 넣으며 아래 줄을 반복합니다.

````python
#> range(5)의 수를 하나씩 i에 넣으며 출력합니다
# for ... in ...:

for i in range(5):
    print(i)

# for range의 범위만큼 반복하면서 하나의 숫자들을 i에 저장:
#     i = 0
#     print(i) # => 0
#     i = 1
#     print(i) # => 1
#     ...
#     i = 4
#     print(i) # => 4
#     i ?
#     끝
````

| 형태 | 만드는 수 |
| --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 (끝 수는 빠짐) |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (2씩) |

````python
#> 1부터 5까지 반복하며 "1번째 반복" 모양으로 출력합니다 (f-string)
for i in range(5):
    print(f'{i}번째 반복')
````

문자열도 `for`로 한 글자씩 꺼낼 수 있습니다.

````python
#> "파이썬"의 글자를 하나씩 ch에 넣으며 출력합니다
for ch in "파이썬":
    print(ch)
    
````

## 5. 반복과 누적

1부터 10까지 더합니다. 반복 **밖**에서 `total = 0`으로 시작하고, 반복 **안**에서 더해 갑니다.

````python
#> total을 0으로 시작해서 1부터 10까지 더하고, 마지막 줄에 total을 씁니다
total = 0

for i in range(1, 11):
    # print(i)
    total = total + i

    # i = 1
    # total = 0 + 1
    # i = 2
    # total = 1 + 2
    # i = 3
    # ...
    # i = 10
    
print(total)
````

`total = total + i`는 `total += i`로 줄여 쓸 수 있습니다.

````python
#> 같은 계산을 += 로 쓰고, 반복 안에서 i와 total을 출력합니다
total = 0

for i in range(1, 11):
    total += i
    
print(total)
````

> 결과가 이상하면 반복 안에 `print`를 넣어 값을 봅니다.

## 6. `while`

`while 조건:`은 조건이 `True`인 동안 계속 반복합니다. 몇 번 반복할지 모를 때 씁니다.

````python
#> money는 10000, count는 0. money가 3000 이상인 동안 3000씩 빼고 count를 1씩 늘린 뒤, count와 money를 출력합니다
money = 10000
count = 0

while money >= 3000:
    money = money - 3000
    count = count + 1

print(count, money)
````

만 원으로 3,000원짜리를 3번 사고 1,000원이 남았습니다.

### 무한 반복과 Interrupt

조건이 계속 `True`면 반복이 끝나지 않습니다. 칸 왼쪽이 `[*]`로 남아 있으면 아직 실행 중이라는 뜻입니다.

이럴 때는 위쪽 메뉴 **Kernel → Interrupt Kernel**(또는 도구 막대의 ■ 버튼)로 멈춥니다. 커널을 다시 시작하는 것이 아니라 **지금 실행 중인 칸만 멈추는 것**이라 만든 이름은 그대로 남습니다.

````python
#> n을 0으로 시작해서, n이 3보다 작은 동안 n을 출력하고 1씩 늘립니다
n = 0
while n < 3:
    print(n)
    n = n +1
````

### `while True`와 `break`

조건 자리에 `True`를 쓰면 일부러 끝나지 않는 반복을 만듭니다. 안에서 `break`를 만나야 끝납니다. **끝낼 조건이 반복 중간에 정해질 때** 씁니다.

````python
#> balance는 5000. while True 안에서 1200씩 빼다가, balance가 1200보다 작아지면 break합니다. 끝나고 balance를 출력합니다
balance = 5000

while True:
    balance -= 1200

    if balance < 1200:
        break

print(balance)
````

## 7. `break` · `continue`

- `break` — 반복을 **바로 끝냅니다**
- `continue` — 이번 차례의 남은 줄을 건너뛰고 **다음 차례로** 갑니다

````python
#> 1부터 10까지 반복하며 출력하다가, i가 5가 되면 break로 끝냅니다
for i in range(1, 11):
    print(i)
    if i == 5:
        break
````

````python
#> 1부터 10까지 반복하며, 짝수면 continue로 건너뛰고 홀수만 출력합니다
for i in range(1, 11):
    # if i % 2 == 1:
    #     print(i)
    if i % 2 == 0:
        continue
    print(i)
````

`i % 2 == 0`은 "2로 나눈 나머지가 0", 곧 **짝수**인지 묻는 조건입니다.

`input()`은 실행 중에 **키보드로 값을 받습니다.** 칸을 실행하면 칸 아래에 입력 칸이 나오고, 글을 쓰고 Enter를 누르면 그 글이 **문자열**로 돌아옵니다.

````python
#> input()으로 이름을 받아 name에 넣고, "OO님 안녕하세요"를 출력합니다
name = input()
print(f'{name}님 안녕하세요')
````

받은 값은 늘 문자열입니다. 숫자로 계산하려면 01장의 `int()`로 바꿉니다.

````python
#> 나이를 input()으로 받아 int()로 바꾸고, 20 이상이면 "성인입니다", 아니면 "미성년자입니다"를 출력합니다
age = input('나이를 입력해주세요')
# print(age)
# print(type(age))

if int(age) > 20:
    print('성인')
else:
    print('미성년자')
````

`while True` · `break`와 함께 쓰면 **그만둘 때까지 계속 받는** 프로그램이 됩니다. 1차시 봇이 명령을 받아 답하던 것과 같은 모양입니다.

````python
#> while True 안에서 명령을 input()으로 받습니다. "q"면 break, 
# "/lunch"면 "오늘은 라면", 그 밖에는 "모르는 명령입니다"를 출력합니다
while True:
    text = input('명령: ')
    if text == 'q':
        break
    elif text == '/lunch':
        print('오늘은 라면')
    else:
        print('모르는 명령어입니다.')
````

> 입력 칸이 떠 있는 동안 칸 번호는 `[*]`입니다. 입력하지 않고 다른 칸을 실행하면 기다리기만 합니다. 입력 칸에 값을 넣거나 Kernel → Interrupt Kernel로 멈춥니다.

## 정리

| 할 일 | 방법 |
| --- | --- |
| 조건이 참일 때만 | `if 조건:` + 들여쓰기 |
| 여러 갈래 | `if` → `elif` → `else` (처음 맞는 하나만) · `if` 안의 `if` |
| 값 하나 고르기 | `참일 때 값 if 조건 else 거짓일 때 값` |
| 조건 묶기 | `and` · `or` · `not` · `값 or 기본값` · `20 <= x < 30` |
| 안에 있는지 · 값이 없는지 | `in` · `not in` · `is None` |
| 빈 블록 자리 | `pass` |
| 정해진 횟수 반복 | `for i in range(…):` |
| 조건이 참인 동안 반복 | `while 조건:` |
| 반복 끝내기 · 건너뛰기 | `break` · `continue` |
| 번호와 같이 꺼내기 | `for i, x in enumerate(…):` |
| 끝낼 때를 안에서 정하는 반복 | `while True:` + `break` |
| 키보드로 값 받기 | `input()` — 늘 문자열 |
| 끝나지 않는 칸 멈추기 | Kernel → Interrupt Kernel |
