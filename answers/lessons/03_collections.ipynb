## 2. 슬라이싱

`[시작:끝]`으로 **여러 개를 잘라** 꺼냅니다. `range()`처럼 끝 번호는 빠집니다. 리스트와 문자열에서 똑같이 씁니다.

| 쓰는 모양 | 뜻 |
| --- | --- |
| `[1:3]` | 1번부터 3번 앞까지 |
| `[:2]` | 처음부터 2번 앞까지 |
| `[2:]` | 2번부터 끝까지 |
| `[::-1]` | 거꾸로 전부 |

````python
#> menus의 1번부터 3번 앞까지 잘라 봅니다
menus[1:3]
````

````python
#> menus의 처음 두 개를 잘라 봅니다
# menus[0:2]
menus[:2] # => 제일 앞이 시작점이면 0 생략 가능
menus[2:] # => 제일 끝이 끝나는점이면 -1 생략 가능
````

````python
#> word("python")의 앞 세 글자와, 거꾸로 뒤집은 글자를 print 한 번으로 출력합니다
word = 'python'

print(word[0:3])
print(word[::-1])
````

### 리스트 메서드

리스트 이름 뒤에 `.`을 찍고 쓰는 기능을 **메서드**라고 합니다.

| 메서드 | 하는 일 |

| `.append(값)` | 끝에 하나 더하기 |

### `sort()`와 `sorted()`

- sort() : 원본정렬
- sorted() : 정렬한 결과만 반환 (원본은 그대로)

````python
#> scores에 [70, 95, 82]를 넣고, sorted(scores)의 결과와 scores를 print 한 번으로 출력합니다
scores = [95, 5, 20]
print(scores)
print(sorted(scores))
print(scores)
````

````python
#> scores.sort()로 scores 자체를 정렬하고 확인합니다
print(scores.sort())
print(scores)
````

| | `sorted(리스트)` | `리스트.sort()` |
| --- | --- | --- |
| 원래 리스트 | 그대로 | 바뀜 |
| 돌려주는 것 | 정렬된 새 리스트 | `None` |

큰 것부터 정렬하려면 `reverse=True`를 넣습니다.

````python
#> scores를 큰 것부터 정렬한 새 리스트를 만듭니다
scores.sort(reverse = True)
print(scores)
````

## 4. 리스트와 반복

02장에서 문자열을 `for`로 한 글자씩 꺼냈습니다. 리스트는 **값 하나씩** 꺼냅니다.

````python
#> menus = ["라면", "햄버거", "마라탕", "샐러드"]로 다시 만들고, for로 메뉴를 하나씩 출력합니다
menus = ["라면", "햄버거", "마라탕", "샐러드"]

for menu in menus:
    print(menu)    
````

````python
#> enumerate로 번호(1부터)와 메뉴를 같이 출력합니다. enumerate(menus, 1)을 씁니다
for i, menu in enumerate(menus):
    print(i, menu)
````

`in`으로 리스트 안에 값이 있는지 묻습니다.

````python
#> "마라탕"이 menus 안에 있는지 확인합니다
'마라탕' in menus
# 'a' in 'apple'
````

숫자 리스트는 합 · 가장 큰 값 · 가장 작은 값을 바로 구할 수 있습니다.

````python
#> scores = [70, 95, 82]로 다시 만들고, 합 · 최댓값 · 최솟값을 print 한 번으로 출력합니다
scores = [70, 95, 82]

print(max(scores), min(scores), sum(scores))
````

빈 리스트에서 시작해 반복하며 채울 수도 있습니다.

````python
#> passed를 빈 리스트로 시작해서, scores 가운데 80 이상만 append하고 확인합니다
passed = []
scores = [70, 95, 82]

for score in scores:
    if score >= 80:
        passed.append(score)

print(passed)
````

## 6. 딕셔너리

## 12. `map()`

`map(함수, 리스트)`는 리스트의 값마다 함수를 적용합니다. 결과는 `list()`로 감싸서 봅니다.

````python
#> texts = ["10", "20", "30"]. map(int, texts)로 모두 정수로 바꾼 리스트를 만듭니다
texts = ["10", "20", "30"]
result = []

for text in texts:
    result.append(int(text))
print(result)
````

````python
# texts = [int("10"), int("20"), int("30")]
list(map(int, texts))
````

같은 일을 컴프리헨션으로도 할 수 있습니다: `[int(t) for t in texts]`.

## AI 코드 읽기: 두 풀이 비교

같은 과제를 **두 가지 방법으로** 받아 비교합니다. Claude Code에 아래 문장을 그대로 보냅니다.

> 03장까지 배운 것만 써서, scores = {"민수": 80, "지은": 95, "현우": 72, "서연": 88}에서 평균 이상인 학생 이름 리스트를 만드는 코드를 두 가지로 보여 줘. 하나는 for와 append, 하나는 컴프리헨션으로. 함수는 쓰지 말고, 주석 · 설명 · 실행 결과 없이 코드만 보여 주고, 파일은 만들지 마.

**① for와 append로 쓴 코드를 붙여 넣습니다.**

````python
#> for · append 풀이를 붙여 넣습니다scores = {"민수": 80, "지은": 95, "현우": 72, "서연": 88}
total = 0
for score in scores.values():
  total = total + score
average = total / len(scores)

result = []
for name, score in scores.items():
  if score >= average:
      result.append(name)

print(result)
````

**② 컴프리헨션으로 쓴 코드를 붙여 넣습니다.**

````python
#> 컴프리헨션 풀이를 붙여 넣습니다
scores = {"민수": 80, "지은": 95, "현우": 72, "서연": 88}

total = 0
for score in scores.values():
  total = total + score
average = total / len(scores)

result = [name for name, score in scores.items() if score >= average]

print(result)
````

## 정리

| 종류 | 만들기 | 꺼내기 | 바꿀 수 있나 | 언제 |
| --- | --- | --- | --- | --- |
| 리스트 | `[1, 2, 3]` | `[번호]` · `[시작:끝]` | 있음 | 순서 있는 값 여러 개 |
| 튜플 | `(1, 2)` | `[번호]` · 풀기 `x, y = t` | 없음 | 바뀌면 안 되는 묶음 |
| 딕셔너리 | `{"키": 값}` | `[키]` · `.get(키)` | 있음 | 이름표로 찾는 값 |
| 셋 | `{1, 2}` · `set(리스트)` | 꺼낼 수 없음 · `in` | 있음 | 중복 없애기 · 비교 |

| 할 일 | 방법 |
| --- | --- |
| 잘라 꺼내기 | `[시작:끝]` · 거꾸로 `[::-1]` (리스트 · 문자열) |
| 정렬 | `sorted()` (새 리스트) · `.sort()` (그 자체) |
| 짝짓기 · 거꾸로 | `zip()` · `reversed()` |
| 한 줄로 리스트 만들기 | `[값 for x in 리스트 if 조건]` |
| 복사 | `.copy()` — `b = a`는 이름만 하나 더 |

| 에러 | 뜻 |
| --- | --- |
| `IndexError` | 번호가 범위를 벗어남 |
| `KeyError` | 그런 키가 없음 |
| `TypeError: … does not support item assignment` | 바꿀 수 없는 값(문자열 · 튜플)을 바꾸려 함 |
