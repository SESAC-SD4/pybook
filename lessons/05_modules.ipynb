## 5. `if __name__ == "__main__":`

**기본 구조**

```
def 함수이름(매개변수):
    실행할 코드


if __name__ == "__main__":
    직접 실행할 때만 돌릴 코드
```

모듈 파일에 시험 삼아 실행해 보는 코드를 넣어 두면, `import`할 때마다 그 코드도 실행됩니다. 이것을 막는 줄이 `if __name__ == "__main__":`입니다.

````python
#> calc를 import하고, calc.add(10, 20)을 불러 봅니다
````

`import`할 때는 `if` 안의 `print`가 실행되지 않았습니다. 이번에는 파일을 **직접 실행**합니다.

## 정리

| 할 일 | 방법 |
| --- | --- |
| 모듈 가져오기 | `import math` · `from math import ceil` · `from datetime import date, timedelta` |
| 표준 라이브러리 | `math` · `random`(`choice` · `randint` · `sample`) · `datetime`(`date` · `timedelta` · `strftime`) |
| 내 모듈 | 같은 폴더에 `이름.py` → `import 이름` |
| 모듈 파일을 고친 뒤 | 커널 다시 시작 → 다시 `import` |
| 직접 실행할 때만 | `if __name__ == "__main__":` |
| 패키지 | 폴더 + `__init__.py` → `from 폴더 import 모듈` |
| 설치가 필요한 라이브러리 | `uv add` (이 저장소에서는 강사가 넣어 둠 → `uv sync`) |

| 에러 | 뜻 |
| --- | --- |
| `NameError: name 'math' is not defined` | `import`를 하지 않았다 |
| `ModuleNotFoundError: No module named '…'` | 그런 모듈이 없다. 파일 이름 · 위치 · 설치를 확인 |
