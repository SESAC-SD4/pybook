## 1. 칸 두 종류

노트북은 **칸(cell)** 으로 이루어져 있습니다.

| 칸 | 무엇 | 실행하면 |
| --- | --- | --- |
| 마크다운 칸 | 설명 글 (지금 읽는 이 칸) | 글이 보기 좋게 바뀐다 |
| 코드 칸 | Python 코드 | 코드가 실행되고 결과가 칸 아래에 나온다 |

칸을 실행하는 키

| 키 | 실행한 뒤 |
| --- | --- |
| **Shift + Enter** | 다음 칸으로 넘어간다 |
| **Ctrl + Enter** (맥은 Cmd + Enter) | 그 칸에 머문다 |
| **Alt + Enter** (맥은 Option + Enter) | 아래에 새 칸을 만든다 |

## 정리

| 할 일 | 방법 |
| --- | --- |
| 칸 실행 | Shift + Enter (다음 칸으로) · Ctrl + Enter, 맥은 Cmd + Enter (그 칸에 머묾) · Alt + Enter, 맥은 Option + Enter (아래에 새 칸) |
| 결과 보기 | 마지막 줄의 값 · 여러 개면 `print` |
| 결과가 이상할 때 | 칸 왼쪽 번호로 실행 순서 확인 |
| 처음부터 다시 | 커널 다시 시작 → 위에서부터 실행 |
| 에러 | 마지막 줄부터 읽는다 |
