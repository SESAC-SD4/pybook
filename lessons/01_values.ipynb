| 쓰는 방식 | 예 | 어디서 |
| --- | --- | --- |
| 스네이크 케이스 | `user_name`, `total_price` | Python 변수 · 함수 |
| 카멜 케이스 | `userName`, `totalPrice` | 자바스크립트 · 자바 |

## 7. f-string

따옴표 앞에 `f`를 붙이면, 문자열 안의 `{이름}` 자리에 그 값이 들어갑니다. `str()`로 바꾸지 않아도 됩니다.

```python
#> name과 age를 만들고, f-string으로 "홍길동님은 25살입니다"를 만듭니다
```

`{}` 안에는 계산식도 넣을 수 있습니다.

```python
#> {} 안에 age + 1을 넣어 "내년에는 26살입니다"를 만듭니다
```

### 예전 방식 두 가지

f-string은 Python 3.6부터 생겼습니다. 그 전에 쓰던 방식이 오래된 코드와 인터넷 예제에 아직 많이 남아 있습니다. **읽을 수 있으면 됩니다.** 새로 쓸 때는 f-string을 씁니다.

`%` 방식 — 자리에 `%s`(문자열) · `%d`(정수)를 두고, 뒤에 `%`와 값을 씁니다.

```python
#> %s · %d 자리에 name과 age를 넣어 같은 문장을 만듭니다
```

`.format()` 방식 — 자리에 `{}`를 두고, 뒤에 `.format(값, 값)`을 붙입니다.

```python
#> {} 자리에 .format()으로 name과 age를 넣어 같은 문장을 만듭니다
```

| 방식 | 쓰는 모양 |
| --- | --- |
| f-string | `f"{name}님은 {age}살입니다"` |
| `.format()` | `"{}님은 {}살입니다".format(name, age)` |
| `%` | `"%s님은 %d살입니다" % (name, age)` |

## 정리

| 할 일 | 방법 |
| --- | --- |
| 값의 종류 보기 | `type(값)` |
| 따옴표 안의 특별한 글자 | `\n` 줄 바꿈 · `\t` 탭 · `\'` 따옴표 · 여러 줄은 `"""` |
| 이름 붙이기 | `이름 = 값` (오른쪽을 먼저 계산) · 스네이크 케이스 · 예약어는 못 씀 |
| 몫 · 나머지 | `//` · `%` |
| 자기 자신 바꾸기 | `+=` `-=` `*=` … |
| 계산 순서 | `**` → `*` `/` → `+` `-`, 먼저 할 것은 괄호 |
| 글자 수 | `len(문자열)` |
| 종류 바꾸기 | `int()` · `float()` · `str()` |
| 문자열에 값 넣기 | `f"{이름}"` (예전 방식: `.format()` · `%`) |
| 비교 | `==` `!=` `>` `<` `>=` `<=` → `True` / `False` |
| 참 · 거짓으로 바꾸기 | `bool()` — `0` · `""` · `None`은 `False` |

| 에러 | 뜻 |
| --- | --- |
| `NameError` | 만든 적 없는 이름 |
| `SyntaxError` | 문법이 틀림 (실행 전에 멈춤) |
| `TypeError` | 종류가 맞지 않음 |
| `ValueError` | 종류는 맞는데 값이 맞지 않음 |
