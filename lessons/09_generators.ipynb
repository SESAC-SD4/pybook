## 6. 데코레이터 만들기

**기본 구조**

```
def 데코레이터이름(func):
    def wrapper(*args, **kwargs):
        앞에서 할 일
        result = func(*args, **kwargs)
        뒤에서 할 일
        return result
    return wrapper

@데코레이터이름
def 함수이름(매개변수):
    실행할 코드
```

여러 함수에 **같은 앞뒤 처리**(실행 전후 출력, 시간 재기 등)를 붙이고 싶을 때가 있습니다. 함수마다 같은 줄을 넣지 않고, **함수를 받아 감싼 함수를 돌려주는 함수**를 하나 만듭니다. 이것이 **데코레이터**입니다.

````python
#> 함수 func를 받아, 부르기 전에 "시작: 함수이름", 부른 뒤 "끝: 함수이름"을 출력하는 wrapper를 돌려주는 log(func)를 만듭니다
````

- `func.__name__` — 함수의 이름(`"add"`)입니다
- `wrapper(*args, **kwargs)` — 감쌀 함수가 어떤 인수를 받든 그대로 넘깁니다(04장 6절)
- `result = func(…)` · `return result` — 감싼 함수의 결과를 그대로 돌려줍니다

````python
#> 두 수를 더하는 add(a, b)를 만들고, add = log(add)로 감싼 뒤 add(2, 3)을 부릅니다
````

`@log`를 함수 위에 쓰면 `add = log(add)`와 같습니다.

````python
#> @log를 붙인 multiply(a, b)를 만들고 multiply(4, 5)를 부릅니다
````

## 정리

| 할 일 | 방법 |
| --- | --- |
| 하나씩 내주는 함수 | `yield` — 부르면 제너레이터, `next()` · `for`로 꺼낸다 |
| 다 꺼낸 뒤 | `StopIteration` (`for`는 조용히 끝남) · 다시 쓰려면 다시 부른다 |
| 한 줄로 | `(값 for x in … if 조건)` · 함수에 넣을 때는 `sum(x for x in …)` |
| 함수를 값으로 | 이름에 넣기 · 인수로 넘기기 · 함수에서 돌려주기 |
| 앞뒤 처리 붙이기 | 데코레이터 `def deco(func): def wrapper(*args, **kwargs): … return wrapper` · `@deco` |
