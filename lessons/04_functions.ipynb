## 1. 함수 만들기

점수 평균을 두 번 구하면 이렇게 됩니다.

````python
scores_a = [80, 90, 70]
average_a = sum(scores_a) / len(scores_a)

scores_b = [60, 75, 95, 85]
average_b = sum(scores_b) / len(scores_b)

print(average_a, average_b)
````

같은 계산을 `def`로 묶어 이름을 붙입니다.

**기본 구조**

```

    실행할 코드
    return 돌려줄 값

함수이름(인수)
```

````python
#> scores를 받아 평균을 return하는 함수 average를 만듭니다
````

만든 함수는 `이름(값)`으로 **부릅니다(호출)**.

````python
#> average를 scores_a와 scores_b로 각각 불러 print 한 번으로 출력합니다
````

- `def` 줄 끝에 콜론, 아래 줄은 들여쓰기 — `if` · `for`와 같습니다
- `scores`처럼 함수가 받는 이름을 **매개변수**, 부를 때 넣는 값(`scores_a`)을 **인수**라고 합니다
- 함수를 만드는 칸은 실행해도 아무것도 나오지 않습니다. **불러야** 실행됩니다

## 5. 스코프

**기본 구조**

```
밖의이름 = 값              # 함수 밖 — 함수 안에서 읽을 수 있다

def 함수이름():
    안의이름 = 값          # 함수 안 — 함수 밖에서는 쓸 수 없다
```

함수 **안**에서 만든 이름은 함수 밖에서 쓸 수 없습니다. 이름이 쓰이는 범위를 **스코프(scope)** 라고 합니다.

````python
#> price를 받아 tax = price * 0.1을 만들고 price + tax를 return하는 함수 with_tax를 만든 뒤, 함수 밖에서 tax를 출력해 봅니다
````

함수 밖에서 만든 이름은 함수 안에서 **읽을 수** 있습니다.

````python
#> 함수 밖에 rate = 0.1을 두고, 그 rate를 읽어 할인 금액을 return하는 함수 discount를 만들어 20000으로 불러 봅니다
````

> 함수 안에서 밖의 이름을 **바꾸는** 방법(`global`)도 있지만 쓰지 않습니다. 필요한 값은 인수로 받고, 결과는 `return`으로 돌려줍니다.

| `dict[str, int]` | 키는 문자열, 값은 정수인 딕셔너리 |
| `str \| None` | 문자열이거나 `None` |

````python
#> scores: list[int]를 받아 float를 돌려주는 average를 타입힌트로 다시 만들고 불러 봅니다
````

## 정리

| 할 일 | 방법 |
| --- | --- |
| 함수 만들기 · 부르기 | `def 이름(매개변수):` · `이름(인수)` |
| 값 돌려주기 | `return` (없으면 `None`) · 여러 값은 튜플 |
| 인수 넣기 | 위치 · 키워드 · 기본값 · `*args` · `**kwargs` · `*` 뒤는 키워드로만 |
| 기본값 | 리스트 대신 `None` → 안에서 새로 만들기 |
| 이름 없는 한 줄 함수 | `lambda x: …` — `sorted(key=…)` · `filter()` |
| 설명 남기기 | 독스트링 `"""…"""` · `help()` |
| 종류 적어 두기 | 타입힌트 `x: int` · `-> str` (검사는 하지 않음) |

| 에러 | 뜻 |
| --- | --- |
| `TypeError: … missing 1 required positional argument` | 필요한 인수를 빼먹음 |
| `TypeError: … takes 1 positional argument but 2 were given` | 위치 인수를 너무 많이 넣음 (`*` 뒤는 키워드로) |
| `NameError` (함수 밖에서) | 함수 안에서 만든 이름 — 스코프 밖 |
