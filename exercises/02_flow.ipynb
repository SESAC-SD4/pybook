1. 빈 칸에 코드를 쓰고 실행합니다
2. 바로 아래 확인 칸을 실행합니다. **아무것도 나오지 않으면 통과**입니다
3. 통과하지 못하면 `AssertionError`와 함께 무엇을 고쳐야 하는지 나옵니다. `위 빈 칸을 먼저 채우고 실행해 주세요`가 나오면 빈 칸을 채워 실행했는지 확인합니다

확인 칸은 고치지 않고 실행만 합니다.

## 문제 1. 홀짝

`number`에 `7`을 넣고, 짝수면 `kind`에 `"짝수"`, 홀수면 `"홀수"`를 넣습니다. `if` · `else`와 `%`를 씁니다.

```python
#> number를 만들고, if · else로 kind를 정합니다
```

```python
# 확인 칸
assert "kind" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert kind == "홀수", "7은 2로 나눈 나머지가 1입니다"
```

## 문제 2. 요금 정하기

나이에 따라 입장료가 다릅니다. `age`에 `8`을 넣고 `fee`를 정합니다.

| 나이 | 입장료 |
| --- | --- |
| 7살 이하 | 0 |
| 8~19살 | 5000 |
| 20살 이상 | 10000 |

```python
#> age를 만들고, if · elif · else로 fee를 정합니다
```

```python
# 확인 칸
assert "fee" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert fee == 5000, "8살은 8~19살 칸입니다. elif의 조건과 순서를 확인해 주세요"
```

## 문제 3. 1부터 100까지 더하기

`for`와 `range()`로 1부터 100까지 더한 값을 `total`에 넣습니다.

```python
#> total을 0으로 시작해서 for로 더해 갑니다
```

```python
# 확인 칸
assert "total" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert total != 4950, "100이 빠졌습니다. range의 끝 수는 포함되지 않습니다"
assert total == 5050, "1부터 100까지의 합은 5050입니다"
```
