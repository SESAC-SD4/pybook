```python
# 준비: 이 칸을 먼저 실행합니다
import asyncio
import time


def slow_square(n):
    time.sleep(0.5)
    return n * n
```

`time.sleep(0.5)`는 0.5초 기다립니다. `slow_square`는 0.5초 기다린 뒤 제곱을 돌려줍니다.

## 문제 1. 비동기로 감싸기

`slow_square`를 `asyncio.to_thread`로 불러 결과를 돌려주는 `async def` 함수를 완성합니다.

```python
#> 함수를 완성합니다
async def square_async(n: int) -> int:
    pass
```

```python
# 써 보기
print(await square_async(4))  # => 16
```

```python
# 확인 칸
assert "square_async" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
result = await square_async(4)
assert result is not None, "await asyncio.to_thread(slow_square, n)의 결과를 return합니다"
assert type(result) == int, "숫자가 아닙니다. 써 보기에 <coroutine object …>가 나오면 asyncio.to_thread 앞에 await를 붙입니다"
assert result == 16, "4의 제곱은 16입니다"
```

```python
#> 함수를 완성합니다
async def squares_async(numbers: list[int]) -> list[int]:
    pass
```

```python
# 써 보기
start = time.time()
print(await squares_async([1, 2, 3]))  # => [1, 4, 9]
print("걸린 시간:", time.time() - start, "초")  # => 0.5초쯤
```
