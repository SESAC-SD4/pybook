0부터 `n`까지(포함) 짝수를 하나씩 `yield`하는 제너레이터 함수를 완성합니다.

```python
#> 함수를 완성합니다
def even_numbers(n: int):
    pass
```

```python
# 써 보기
print(list(even_numbers(6)))  # => [0, 2, 4, 6]
```

```python
# 확인 칸
assert "even_numbers" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
gen = even_numbers(6)
assert gen is not None and hasattr(gen, "__next__"), "return 대신 yield를 써서 제너레이터를 만듭니다"
assert list(gen) == [0, 2, 4, 6], "0부터 6까지(포함) 짝수는 0, 2, 4, 6입니다"
assert list(even_numbers(1)) == [0], "n이 1이면 0 하나입니다"
```

문자열 리스트에서 정수로 바꿀 수 있는 값만 `int`로 바꿔 하나씩 `yield`하는 제너레이터 함수를 완성합니다.

```python
#> 함수를 완성합니다
def only_numbers(texts: list[str]):
    pass
```

```python
# 써 보기
print(list(only_numbers(["1", "둘", "3", ""])))  # => [1, 3]
```

```python
# 확인 칸
assert "only_numbers" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
gen = only_numbers(["1", "둘", "3", ""])
assert gen is not None and hasattr(gen, "__next__"), "yield를 써서 제너레이터를 만듭니다"
try:
    result = list(gen)
except ValueError:
    assert False, "\"둘\" · \"\"은 int로 바꿀 수 없어 ValueError가 납니다. int(…)를 try로 감싸고 except ValueError에서 넘어갑니다"
assert result == [1, 3], "바꿀 수 있는 \"1\"과 \"3\"만 int로 내줍니다. try · except ValueError"
```
