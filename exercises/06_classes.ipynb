## 문제 1. 카운터

속성 `count`가 0에서 시작하고, `increment()`를 부를 때마다 1씩 늘어나는 `Counter` 클래스를 완성합니다.

```python
#> 클래스를 완성합니다
class Counter:
    def __init__(self):
        pass

    def increment(self):
        pass
```

```python
# 써 보기
c = Counter()
c.increment()
c.increment()
print(c.count)  # => 2
```

```python
# 확인 칸
assert "Counter" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
c = Counter()
assert hasattr(c, "count"), "__init__에서 self.count를 만듭니다"
assert c.count == 0, "count는 0에서 시작합니다"
c.increment()
c.increment()
assert c.count == 2, "increment()를 두 번 부르면 2입니다. self.count += 1"
assert Counter().count == 0, "새 인스턴스는 다시 0에서 시작합니다"
```

## 문제 2. 직사각형

가로 `width`, 세로 `height`를 받고, 넓이 `area()`와 둘레 `perimeter()`를 돌려주는 `Rectangle` 클래스를 완성합니다.

```python
#> 클래스를 완성합니다
class Rectangle:
    def __init__(self, width, height):
        pass

    def area(self):
        pass

    def perimeter(self):
        pass
```

```python
# 써 보기
r = Rectangle(3, 4)
print(r.area(), r.perimeter())  # => 12 14
```

```python
# 확인 칸
assert "Rectangle" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
r = Rectangle(3, 4)
assert r.area() is not None, "print가 아니라 return으로 돌려줍니다"
assert r.area() == 12, "3 × 4 = 12"
assert r.perimeter() == 14, "둘레는 2 × (3 + 4) = 14"
assert Rectangle(5, 5).area() == 25, "5 × 5 = 25"
```

## 문제 4. 가입한 회원 수

```python
# 확인 칸
assert "Member" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
start = Member.count
m = Member("민수")
Member("지은")
assert Member.count == start + 2, "만들 때마다 Member.count += 1 (self.count가 아니라 클래스 이름으로)"
assert hasattr(m, "name") and m.name == "민수", "__init__에서 self.name을 넣습니다"
```

## 문제 5. 문자열로 만들기

`"커피,4000"`처럼 이름과 가격을 쉼표로 이은 문자열로 `Product`를 만드는 클래스 메서드 `from_text`를 완성합니다. 가격은 정수로 바꿉니다.

```python
#> 클래스를 완성합니다
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @classmethod
    def from_text(cls, text):
        pass
```

```python
# 써 보기
p = Product.from_text("커피,4000")
print(p.name, p.price)  # => 커피 4000
```

```python
# 확인 칸
assert "Product" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
p = Product.from_text("커피,4000")
assert p is not None, "from_text는 cls(...)로 만든 인스턴스를 return합니다"
assert p.name == "커피", "쉼표 앞이 이름입니다. split(\",\")"
assert p.price == 4000 and type(p.price) == int, "가격은 int()로 정수로 바꿉니다"
```

## 문제 6. 동물 소리

`Animal`의 `speak()`는 `"..."`를 돌려줍니다. `Animal`을 상속한 `Dog`에서 `speak()`를 다시 만들어 `"이름: 멍멍"`을 돌려주게 합니다. `Dog("바둑이").speak()`는 `"바둑이: 멍멍"`입니다.

```python
#> 클래스를 완성합니다
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."


class Dog(Animal):
    pass
```

```python
# 써 보기
print(Dog("바둑이").speak(), Animal("고양이").speak())  # => 바둑이: 멍멍 ...
```

```python
# 확인 칸
assert "Dog" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
d = Dog("바둑이")
assert isinstance(d, Animal), "Dog는 Animal을 상속합니다. class Dog(Animal):"
assert d.speak() == "바둑이: 멍멍", "Dog(\"바둑이\").speak()는 \"바둑이: 멍멍\"입니다. 이름 자리에 self.name을 넣습니다"
assert Animal("고양이").speak() == "...", "Animal의 speak()는 그대로 둡니다"
```

## 문제 7. 장바구니 항목

`@dataclass`로 `Item`을 만듭니다. `name: str`, `price: int`, `qty: int`(기본값 1)를 가지고, `total()`은 `price * qty`를 돌려줍니다.

```python
#> 클래스를 완성합니다
from dataclasses import dataclass


@dataclass
class Item:
    pass
```

```python
# 써 보기
item = Item("라면", 4000, 2)
print(item, item.total())  # => Item(name='라면', price=4000, qty=2) 8000
```

```python
# 확인 칸
assert "Item" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert hasattr(Item, "total"), "Item 안에 name · price · qty 속성과 total() 메서드를 만듭니다"
assert Item("라면", 4000, 2).total() == 8000, "total()은 price * qty입니다"
assert Item("김밥", 3000).qty == 1, "qty의 기본값은 1입니다. qty: int = 1"
assert Item("라면", 4000) == Item("라면", 4000), "@dataclass면 같은 값끼리 ==가 True입니다"
```

### 참고 코드: 포켓몬 배틀

| 클래스 | 속성 (가진 것) | 메서드 (하는 일) |
| --- | --- | --- |
| `Skill` | `name`, `power` | — |
| `Pokemon` | `name`, `level`, `hp`, `skills`(`Skill` 리스트) | 상대 `hp` 줄이기 `attack(other)` · 살아 있는지 `is_alive()` · `__str__` |

실행할 때마다 결과가 다릅니다(`random`).

```python
import random


class Skill:
    def __init__(self, name, power):
        self.name = name
        self.power = power


class Pokemon:
    def __init__(self, name, level, skills):
        self.name = name
        self.level = level
        self.hp = level * 10
        self.skills = skills

    def __str__(self):
        return f"{self.name} (HP {self.hp})"

    def is_alive(self):
        return self.hp > 0

    def attack(self, other):
        skill = random.choice(self.skills)
        damage = skill.power + random.randint(0, self.level)
        other.hp = max(other.hp - damage, 0)
        print(f"{self.name}의 {skill.name}! {other.name}에게 {damage} 데미지")


kobugi = Pokemon("꼬부기", 5, [Skill("물대포", 8), Skill("몸통박치기", 5)])
pairi = Pokemon("파이리", 5, [Skill("불꽃세례", 9), Skill("할퀴기", 4)])

attacker, defender = kobugi, pairi
while kobugi.is_alive() and pairi.is_alive():
    attacker.attack(defender)
    print(kobugi, "/", pairi)
    attacker, defender = defender, attacker

winner = kobugi if kobugi.is_alive() else pairi
print(f"{winner.name} 승리!")
```
