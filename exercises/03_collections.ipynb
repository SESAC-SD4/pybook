# 03 연습

혼자 푸는 문제입니다. 문제마다 **빈 칸**과 **확인 칸**이 있습니다.

1. 빈 칸에 코드를 쓰고 실행합니다
2. 바로 아래 확인 칸을 실행합니다. **아무것도 나오지 않으면 통과**입니다
3. 통과하지 못하면 `AssertionError`와 함께 무엇을 고쳐야 하는지 나옵니다. `위 빈 칸을 먼저 채우고 실행해 주세요`가 나오면 빈 칸을 채워 실행했는지 확인합니다

확인 칸은 고치지 않고 실행만 합니다.

## 문제 1. 인덱스

`fruits`에 `["사과", "바나나", "포도", "딸기"]`를 넣고, 두 번째 과일을 `second`에, 마지막 과일을 `last`에 넣습니다. `last`는 음수 인덱스로 꺼냅니다.

```python
#> fruits를 만들고, second와 last를 꺼냅니다
```

```python
# 확인 칸
assert "second" in globals() and "last" in globals(), "위 빈 칸을 먼저 채우고 실행해 주세요"
assert second != "사과", "인덱스는 0부터 시작합니다. 두 번째는 1번입니다"
assert second == "바나나", "두 번째 과일은 바나나입니다"
assert last == "딸기", "마지막 값은 [-1]로 꺼냅니다"
```
